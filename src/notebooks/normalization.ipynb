```python
from pathlib import Path

from dataset import UnlabeledDetectionDataset, DetectionDataset
from training import explore_dataset


dataset_path = Path("../data_det_public/data_det_public")
df = explore_dataset(dataset_path)
dataset = UnlabeledDetectionDataset(df)
```

```python
import torch
from torch import Tensor


def compute_mean_std(dataset: DetectionDataset) -> tuple[Tensor, Tensor]:
    sm = torch.zeros(3, dtype=torch.float64)
    sq_sm = torch.zeros(3, dtype=torch.float64)
    total = 0

    for i in range(len(dataset)):
        im, _ = dataset[i]
        im = im.float()
        sm += torch.sum(im, dim=[1, 2])
        sq_sm += torch.sum(im**2, dim=[1, 2])
        total += im.shape[1] * im.shape[2]

    mean = sm / total
    var = (sq_sm / total) - (mean**2)
    std = torch.sqrt(var)
    return mean, std
```

```python
m, s = compute_mean_std(dataset)
```
